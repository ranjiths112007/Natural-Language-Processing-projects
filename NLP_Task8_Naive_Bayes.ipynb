```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Sample dataset
data = {
    'text': [
        'Win a free iPhone now',
        'Congratulations you won a cash prize',
        'Claim your free reward today',
        'Get free money by clicking this link',
        'You have won a lottery prize',
        'Exclusive offer buy now and get discount',
        'Free entry in our weekly contest',
        'Urgent claim your reward now',
        'Meeting is scheduled for tomorrow',
        'Please send me the project report',
        'Can you call me after class',
        'The exam starts tomorrow',
        'Please bring your notebook to class',
        'Let us meet for lunch today',
        'Your assignment is due Friday',
        'I will send the files this evening',
        'Are you coming to college today',
        'The project presentation went well',
        'Please check the email I sent you',
        'We have a team meeting at five'
    ],
    'label': [
        'spam','spam','spam','spam','spam','spam','spam','spam',
        'ham','ham','ham','ham','ham','ham','ham','ham','ham','ham','ham','ham'
    ]
}

df = pd.DataFrame(data)
```

```python
# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    df['text'], df['label'], test_size=0.25, random_state=42, stratify=df['label']
)

# Convert text to TF-IDF features
vectorizer = TfidfVectorizer(stop_words='english')
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

# Train Naive Bayes model
model = MultinomialNB()
model.fit(X_train_tfidf, y_train)

# Evaluate
y_pred = model.predict(X_test_tfidf)
print('Accuracy:', accuracy_score(y_test, y_pred))
print('\nClassification Report:\n')
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
# Enter your own text here
user_input = input('Enter a message: ')

input_tfidf = vectorizer.transform([user_input])
prediction = model.predict(input_tfidf)[0]

print('\nInput:', user_input)
print('Prediction:', prediction.upper())
```
