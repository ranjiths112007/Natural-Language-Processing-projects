# NLP Task 7 - Doc2Vec

This notebook demonstrates how to train a Doc2Vec model and generate document vectors.

```python
# Install dependencies in a fresh notebook environment if needed:
# %pip install -U gensim nltk scipy

import nltk

nltk.download('punkt', quiet=True)
try:
    nltk.download('punkt_tab', quiet=True)
except Exception:
    # Older NLTK versions do not provide punkt_tab.
    pass

from gensim.models.doc2vec import Doc2Vec, TaggedDocument
from nltk.tokenize import word_tokenize

print('Libraries imported successfully!')
```

```python
data = [
    'This is the first document',
    'This is the second document',
    'This is the third document',
    'This is the fourth document'
]

if not data:
    raise ValueError('The document collection is empty.')

print(f'{len(data)} documents loaded.')
```

```python
tagged_data = [
    TaggedDocument(
        words=word_tokenize(doc.lower()),
        tags=[str(i)]
    )
    for i, doc in enumerate(data)
]

print('Tagged documents created successfully!')
```

```python
model = Doc2Vec(
    vector_size=20,
    min_count=1,
    epochs=50,
    workers=1,
    seed=42
)

model.build_vocab(tagged_data)
model.train(
    tagged_data,
    total_examples=model.corpus_count,
    epochs=model.epochs
)

print('Doc2Vec model trained successfully!')
```

```python
document_vectors = [
    model.infer_vector(word_tokenize(doc.lower()))
    for doc in data
]

print('Document vectors generated successfully!')
```

```python
print('Vector size:', model.vector_size)

for i, vector in enumerate(document_vectors):
    print(f'Document {i + 1} vector length:', len(vector))
```

```python
# Display the learned vector for each document.
for i, vector in enumerate(document_vectors):
    print(f'Document {i + 1}:')
    print(vector)
```
