# NLP Task 8: Text Classification using Naive Bayes

This notebook demonstrates text classification using **TF-IDF** features and a **Multinomial Naive Bayes** classifier.


## 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay
```

## 2. Create a Sample Text Dataset

The dataset contains short messages classified as **spam** or **ham**.

```python
data = {
    'text': [
        'Win a free iPhone now click here',
        'Congratulations you won a cash prize',
        'Claim your free reward today',
        'Get free money by clicking this link',
        'You have won a lottery prize',
        'Exclusive offer buy now and get a discount',
        'Free entry in our weekly contest',
        'Urgent claim your reward now',
        'Meeting is scheduled for tomorrow morning',
        'Please send me the project report',
        'Can you call me after class',
        'The exam starts at ten tomorrow',
        'Please bring your notebook to class',
        'Let us meet for lunch today',
        'Your assignment submission is due Friday',
        'I will send the files this evening',
        'Are you coming to college today',
        'The project presentation went well',
        'Please check the email I sent you',
        'We have a team meeting at five'
    ],
    'label': [
        'spam', 'spam', 'spam', 'spam', 'spam', 'spam', 'spam', 'spam',
        'ham', 'ham', 'ham', 'ham', 'ham', 'ham', 'ham', 'ham',
        'ham', 'ham', 'ham', 'ham'
    ]
}

df = pd.DataFrame(data)
df.head()
```

## 3. Check the Dataset

```python
print('Dataset shape:', df.shape)
print('\nClass distribution:')
print(df['label'].value_counts())
```

## 4. Split Data into Training and Testing Sets

```python
X = df['text']
y = df['label']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

print('Training samples:', len(X_train))
print('Testing samples:', len(X_test))
```

## 5. Convert Text into TF-IDF Features

```python
vectorizer = TfidfVectorizer(stop_words='english')

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print('Training feature matrix shape:', X_train_tfidf.shape)
print('Testing feature matrix shape:', X_test_tfidf.shape)
```

## 6. Train the Naive Bayes Classifier

```python
model = MultinomialNB()
model.fit(X_train_tfidf, y_train)

print('Naive Bayes model trained successfully!')
```

## 7. Make Predictions

```python
y_pred = model.predict(X_test_tfidf)

results = pd.DataFrame({
    'Text': X_test.values,
    'Actual': y_test.values,
    'Predicted': y_pred
})
results
```

## 8. Evaluate the Model

```python
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy:.2%}')

print('\nClassification Report:')
print(classification_report(y_test, y_pred, zero_division=0))
```

## 9. Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred, labels=['ham', 'spam'])
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['ham', 'spam'])
disp.plot()
plt.title('Naive Bayes - Confusion Matrix')
plt.show()
```

## 10. Test with New Text

The trained model can classify completely new messages.

```python
new_texts = [
    'Congratulations you won free money',
    'Please send the assignment report tomorrow',
    'Claim your free prize now',
    'Can we meet after class'
]

new_text_tfidf = vectorizer.transform(new_texts)
new_predictions = model.predict(new_text_tfidf)

for text, prediction in zip(new_texts, new_predictions):
    print(f'Text: {text}')
    print(f'Prediction: {prediction}\n')
```

## Conclusion

Naive Bayes is a simple and effective machine learning algorithm for text classification. In this task, TF-IDF converts text into numerical features and Multinomial Naive Bayes uses those features to classify messages as spam or ham.
