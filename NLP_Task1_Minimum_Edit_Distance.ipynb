```python
def minimum_edit_distance(source, target):
    rows, cols = len(source) + 1, len(target) + 1
    dp = [[0] * cols for _ in range(rows)]

    for i in range(rows):
        dp[i][0] = i
    for j in range(cols):
        dp[0][j] = j

    for i in range(1, rows):
        for j in range(1, cols):
            if source[i - 1] == target[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]
            else:
                dp[i][j] = 1 + min(
                    dp[i][j - 1],
                    dp[i - 1][j],
                    dp[i - 1][j - 1]
                )
    return dp
```

```python
source = 'kitten'
target = 'sitting'
dp = minimum_edit_distance(source, target)

print(f'Source: {source}')
print(f'Target: {target}')
print(f'Minimum Edit Distance: {dp[-1][-1]}')
print('\nDP Matrix:')
for row in dp:
    print(row)
```
