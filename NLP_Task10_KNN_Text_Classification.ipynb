```python
# NLP Task: KNN text classification
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Libraries loaded successfully.")
```

```python
# Small, easy-to-read dataset for two text categories
data = {
    "text": [
        "The team won the football match last night",
        "He scored a goal in the final minute",
        "The cricket player hit a century today",
        "Our team is training hard for the championship",
        "The referee gave a red card in the match",
        "The basketball team practiced free throws",
        "New smartphone released with faster processor",
        "The laptop comes with an upgraded graphics card",
        "Artificial intelligence is transforming software development",
        "The new update improves battery life on phones",
        "Developers released a new version of the app",
        "The computer crashed due to a software bug"
    ],
    "label": [
        "sports", "sports", "sports", "sports", "sports", "sports",
        "technology", "technology", "technology", "technology", "technology", "technology"
    ]
}

df = pd.DataFrame(data)

print("Dataset preview:")
print(df.to_string(index=False))
print("\nClass counts:")
print(df["label"].value_counts())
```

```python
# Convert text into TF-IDF features
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df["text"])
y = df["label"]

print("TF-IDF feature matrix shape:", X.shape)
print("Number of vocabulary terms:", len(vectorizer.vocabulary_))
print("First few terms:", list(vectorizer.vocabulary_)[:10])
```

```python
# Keep the test split reproducible so the result is easy to check
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
# KNN works on the TF-IDF vectors using cosine distance
knn = KNeighborsClassifier(n_neighbors=3, metric="cosine")
knn.fit(X_train, y_train)

print("KNN model trained successfully.")
```

```python
# Check how the model performs on unseen text
y_pred = knn.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Actual labels:     ", list(y_test))
print("Predicted labels:  ", list(y_pred))
print("\nAccuracy:", round(accuracy, 2))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
# Try the trained model on a few new sentences
new_texts = [
    "The striker missed an easy goal",
    "The company launched a new AI chip",
    "She bought a new gaming laptop",
    "The coach called a timeout during the match"
]

new_vectors = vectorizer.transform(new_texts)
predictions = knn.predict(new_vectors)

print("New text predictions:")
for text, label in zip(new_texts, predictions):
    print(f"{text}  ->  {label}")

print("\nPrediction completed successfully.")
```
